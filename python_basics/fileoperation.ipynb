```python
#### Read a whole file 

with open('example.txt','r') as file:
    content = file.read()
    print(content)

```

```python
## Read a file line by line 

with open('example.txt','r') as file:
    for line in file:
        print(line.strip())    #strip() removes the newline character
```

```python
### Writing a file (Overwriting)

with open('example.txt','w') as file:
    file.write('Noiceeeee!!!!\n')
    file.write('This is a new line')
    
```

```python
## Write a file(without overwritting)

with open('example.txt','a') as file:
    file.write('\nI m appending a new line')
```

```python
### Writing a list of lines to a file 

lines = ['First line\n','Second line\n','Third line\n']

with open('example.txt','a') as file:
    file.writelines(lines)
```

```python
### Binary files 
## Writing to a binary file 

data = b'\x00\x01\x02\x03\x04'

with open('example.bin','wb') as file:
    file.write(data)

with open('example.bin','rb') as file:
    content = file.read()
    print(content)  # prints b'\x00\x01\x02\x03\x04'
```

```python
###Read the content from source text file and write to a destination text file 

# Copying a text file 

with open('example.txt','r') as source_file:
    content = source_file.read()

with open('example.txt','w') as dest_file:
    dest_file.write(content)
```

```python
### Read a textfile and count number of lines, words and characters.

def count_text(file_path):
    with open(file_path,'r') as file:
        lines = file.readlines()
        line_count = len(lines)
        word_count = sum(len(line.split()) for line in lines)
        char_count = sum(len(line) for line in lines)
    return line_count,word_count,char_count

file_path = 'example.txt'
lines, words, chars = count_text(file_path)
print(f'Lines: {lines},Words:{words},Chars:{chars}')
```

```python
##Writing and then reading a file 
##w+ mode is used for both writing and reading data


with open('example.txt','w+') as file:
    file.write('Hello, world!')
    file.write('New line\n')

    ## Move the file cursor to the beginning 
    file.seek(0)

    ## Read the content of the file 

    content = file.read()
    print(content)

```
