```python
student = {
    "name":"Kinnari",
    "age": 22,
    "city":"Mumbai"
}

print(student)
```

```python
#accessing 

print(student['name'])


##using get() method

print(student.get('age'))
print(student.get('last name',"NA"))
```

```python
#modify 

student["age"] = 23    #updated value
student["address"]="Dadar"  #added a new key
print(student)
```

```python
del student['address']
print(student)
```

```python
##methods

keys=student.keys()   #prints all keys
print(keys)
values = student.values()    #prints all values
print(values)
```

```python
items = student.items()  #get all key-value pairs
print(items)
```

```python
##shallow copy

student
```

```python
student_copy=student 
print(student)
print(student_copy)
```

```python
student["name"]="kinnari1"
print(student)
print(student_copy)

#here , we can see that after I assign or modify a value in the original dict, it is changing in the copy one as well. To avoid this, we use shallow copy
```

```python

##shallow copy - when we change one variable, it shouldn't impact the other

student_copy1 = student.copy()
print(student_copy1)
print(student)
```

```python
student["name"]="kinnari2"
print(student_copy1)
print(student)
```

```python
#iteration

for keys in student.keys():
    print(keys)
    
```

```python
for key, value in student.items():
    print(f"{key}: {value}")
```

```python
##nested dicts 

students = {
    "student1": {"name":"kinnari","age":23},
    "student2": {"name":"kshitija","age":23},
}

print(students)
```

```python
print(students["student2"]["name"])
```

```python
for student_id, info in students.items():
    print(f"{student_id}:{info}")
    for key,value in info.items():
        print(f"{key}:{value}")
```

```python
#dict comprehension 

squares = {x:x**2 for x in range(5)}
print(squares)  # {0: 0, 1: 1, 2
```

```python
#conditional dict 

even = {x:x**2 for x in range(10) if x%2==0}
print(even)  # {0: 0, 2: 4, 4
```

```python
##practical examples

#use a dict to count freq of elements in list

num = [1,2,2,3,3,3,4,4,4]

freq = {}

for nums in num:
    if nums in freq:
        freq[nums] += 1
    else:
        freq[nums] = 1
print(freq)
    
```

```python
##merge 2 dicts 

dict1 = {"a":1,"b":2}
dict2 = {"b":3,"c":4}

merged = {**dict1,**dict2}  
print(merged)  # Output: {'a': 1, 'b': 3,
```
