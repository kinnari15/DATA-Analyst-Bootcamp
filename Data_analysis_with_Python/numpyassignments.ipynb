```python
array = np.arange(1, 10).reshape((3, 3))
print("Original array:")
print(array)

# Normalize the array
mean = np.mean(array)
std_dev = np.std(array)
normalized_array = (array - mean) / std_dev

print("Normalized array:")
print(normalized_array)
```

```python
import numpy as np
a = np.array([('Kinnari', 23, 48.0), ('Kshitija', 24, 55.0),('Aditi',27,56.0)], 
       dtype=[('name', (np.str_, 10)), ('age', np.int32), ('weight', np.float64)])
             
# Sorting according to the name
b = np.sort(a, order='name')
print('Sorting according to the name', b)

# Sorting according to the age
b = np.sort(a, order='age')
print('\nSorting according to the age', b)
```

```python
import numpy.ma as ma 

arr = np.random.randint(1,21,size=(4,4))
print(arr)
masked = ma.masked_greater(arr, 10)
print(masked)

sum_unmasked = masked.sum()
print(sum_unmasked)

```

```python
arr = np.random.randint(1,21,size=(3,3))
print(arr)

masked = ma.masked_array(arr,np.eye(3,dtype=bool))
print(masked)

unmasked = masked.mean()
print(unmasked)

fill = masked.filled(unmasked)
print(fill)
```
