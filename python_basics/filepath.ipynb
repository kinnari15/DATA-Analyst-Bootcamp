```python
### create new directory 
import os 

new_directory = 'New_folder'
os.mkdir(new_directory)  # create new directory
print(f"Directory {new_directory} created")  # print message to confirm creation
```

```python
###Listing all files and folders 


items = os.listdir('.')
print(items)
```

```python
#### Joining paths 

dir_name = 'folder'
file_name='file.txt'
full_path = os.path.join(os.getcwd(),dir_name,file_name)
print(full_path)
```

```python
path = 'example.txt'
if os.path.exists(path):
    print(f"File {path} exists")
else:
    print(f"File {path} does not exist")
```

```python
###checking if path is file or directory 

path = 'example.txt'
if os.path.isfile(path):
    print(f'{path} is a file')
elif os.path.isdir(path):
    print(f'{path} is a directory')
else:
    print(f'{path} is neither a file nor a directory')
```

```python
#### Getting the absolute path 

relative_path = 'example.txt'
absolute_path = os.path.abspath(relative_path)
print(absolute_path)  # Output: /path/to/example.txt
```
