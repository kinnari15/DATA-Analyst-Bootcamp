```python
##variable length arguments 

def print_numbers(*args):
       #all the arguments
       for num in args:
              print(num)

print_numbers(1,2,3,4,5,6,7,"Kinnari")
```

```python
##keywords argument 

def print_details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_details(name="Kinnari",age="23",country="India")
```

```python
def print_details(*args,**kwargs):

    for val in args:
        print(val)

    for key, value in kwargs.items():
        print(f"{key}: {value}")


print_details(1,2,3,4,name="Kinnari",age="23",country="India")
```

```python
##temp conversion

def convert_temp(temp,unit):
    if unit == 'C':
        return (temp*9/5) + 32
    elif unit == 'F':
        return (temp-32) *5/9
    else:
        return "Invalid unit"
    
convert_temp(25,'C')
    
```

```python
##password checker 

def passcheck(passwd):
    """This checks password is strng or not"""

    if len(passwd)<8:
        return "Password should be at least 8 characters long"
    if not any(char.isdigit() for char in passwd):
        return "Password should have at least one digit"
    if not any(char.islower() for char in passwd):
        return "Password should have at least one lowercase letter"
    if not any(char.isupper() for char in passwd):
        return "Password should have at least one uppercase letter"
    if not any(char in "!@#$%^&*()_+-=" for char in passwd):
        return "Password should have at least one special character"
    return "Password is strong"

passcheck("Kinnari@15")
```

```python
def calculate_total_cost(cart):
    total_cost = 0
    for item in cart:
        total_cost += item['price']*item['quantity']
    
    return total_cost
    


cart = [
    {'name':'Apple',
     'price': 50,
     'quantity': 2},
     {'name':'Banana',
      'price': 20,
      'quantity': 3},
      {'name':'Cherry',
       'price': 30,
       'quantity': 1}
]


total_cost=calculate_total_cost(cart)
print(total_cost)
```

```python
def is_palindrome(s):
    s = s.lower().replace(" ","")
    return s == s[::-1]

print(is_palindrome("A man a plan a canal panama"))
```

```python
def fact(n):
    if n==0:
        return 1
    else:
        return n*fact(n-1)
    
fact(5)
```
