```python
to_do_list=["Read Books","Learn python","Go for a Walk"]

to_do_list.append("Listen to Music")
to_do_list.append("Watch a Movie")

##removing a completed task 

to_do_list.remove("Learn python")

if "Read Books" in to_do_list:
    print("You have a book to read")

print("To do list remaining:")

for task in to_do_list:
    print(f"-{task}")
```

```python
##managing an inventory 

inventory = ["apples","mangoes","bananas","grapes"]

##adding a new item 

inventory.append("oranges")

##removing an out of stock item 

inventory.remove("apples")

item="mangoes"

if item in inventory:
    print(f"{item} is available in the inventory")
else:
    print(f"{item} is out of stock")

```

```python
feedback = ["Amazing","Very satisfied","could be better","Excellent experience"]

feedback.append("Not happy with the service")

positive_feedback_count = sum(1 for comment in feedback if "great" in comment.lower() or "Excellent" in comment.lower())
print(positive_feedback_count)

for comment in feedback:
    print(f"-{comment}")
```
