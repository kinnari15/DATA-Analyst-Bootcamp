```python
##map () --> It applies a function to all the items in the list or any collection
##helpful in list comprehensions

num = [1,2,3,4,5,6]

sqr = lambda x: x**2 
result = map(sqr, num) ##convert to list since only this gives a location
print(list(result))  # Output: [1, 4, 9, 16,

```

```python
nums = [1,2,3,4,5,6]
print(map(sqr,nums))
print(list(map(sqr, nums)))
```

```python
##lambda func with map 

nums = [1,2,3,4,5,6,7,8]
list(map(lambda x:x*x, nums))
```

```python
##multiple iterables 

num1 = [1,2,3]
num2 = [4,5,6]

add = map(lambda a,b:a+b, num1, num2)
print(list(add))  # Output: [5, 7, 9]
```

```python
## convert a list of strings to integers

str1 = ['1','2','3','4','5']
int_num = list(map(int, str1))
print(int_num)  # Output: [1, 2, 3, 4,
```

```python
words = ['apple','banana','cherry']
upper = list(map(str.upper, words ))
print(upper)
```

```python
def get_name(person):
    return person['name']

people = [
    {'name': 'John', 'age': 30},
    {'name': 'Alice', 'age': 25},
    {'name': 'Bob', 'age': 40}

]

list(map(get_name,people))
```
