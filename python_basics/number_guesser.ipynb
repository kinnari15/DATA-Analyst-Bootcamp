```python
import random 

num = random.randint(1,100)
print(num)
guess = None

while guess!=num:
    guess = int(input(("Guess a number: ")))
    print(guess)
    if guess < num:
        print("Too low, try again!")
    elif guess > num:
        print("Too high, try again!")
    else:
        print("Absolutely correct!")
```
