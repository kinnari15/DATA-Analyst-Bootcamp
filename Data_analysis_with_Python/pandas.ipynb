```python
## Series is a 1D array like object that can hold any datatype.

import pandas as pd 

data = [1,2,3,4,5]
series=pd.Series(data)
print(series)
print(type(series))
```

```python
##Creating a series from dictionary 

data = {'a':1,'b':2,'c':3}
series = pd.Series(data)
print(series)
```

```python
data = [10,20,30]
index = ['a','b','c']
pd.Series(data,index=index)
```

```python
## Dataframe - 2D - similiar to an excel

## create a dataframe from a dictionary of list 

data = {
    'Name': ['Kinnari','John','Jack'],
    'Age': [23, 21, 19],
    'City':['Mumbai','New York','Baltimore']

}

df = pd.DataFrame(data)
print(df)
print(type(df))

```

```python
import numpy as np
np.array(df)
```
