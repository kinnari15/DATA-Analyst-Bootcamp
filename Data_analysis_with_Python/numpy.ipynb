```python
arr = np.array([[1,2,3],[4,5,6]])

print("Array:\n",arr)
print("Shape",arr.shape)
print("Number of Dimensions: ", arr.ndim)
print("Array size: ",arr.size)
print("Data type: ", arr.dtype)
print("In bytes", arr.itemsize)
```

```python
### Numpy vectorised operations 

arr1 = np.array([1,2,3,4,5])
arr2 = np.array([6,7,8,9,10])

## Element wise addition
print(arr1+arr2)

print(arr1-arr2)

print(arr1*arr2)

print(arr1/arr2)
```

```python
### Universal functions 

arr = np.array([1,2,3,4,5])
print(np.sqrt(arr))
print(np.exp(arr))
print(np.sin(arr))
print(np.log(arr))
```

```python
### statistical concept ----Normalisation
## to have a mean of zero and standard deviation is 1

data = np.array([1,2,3,4,5])
mean = np.mean(data)
std = np.std(data)

#normalize the data 

norm_data = (data - mean) / std 
print(norm_data)
```

```python
data = np.array([1,2,3,4,5])
mean = np.mean(data)
print(mean)  # Output: 3.0
std = np.std(data)
print(std)  # Output: 1.4142135623730951
median = np.median(data)
print(median)  # Output: 3.0
var = np.var(data)
print(var)
```
