```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5]
y=[1,4,9,16,25]


#create a line plot

plt.plot(x,y)
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.title('Line Plot')
plt.show()
```

```python
x=[1,2,3,4,5]
y=[1,4,9,16,25]

#create customised line plot

plt.plot(x,y,color='red',linestyle='-.',marker='o',linewidth=3,markersize=10)
plt.grid(True)
```

```python
##Multiple plots

x = [1,2,3,4,5]
y1 = [1,4,9,16,25]
y2 = [1,2,3,4,5]

plt.figure(figsize=(9,5))
plt.subplot(1,2,1)
plt.plot(x,y1,'green')
plt.title('Plot 1')

plt.subplot(1,2,2)
plt.plot(y2,x,'red')
plt.title('Plot 2')
```
