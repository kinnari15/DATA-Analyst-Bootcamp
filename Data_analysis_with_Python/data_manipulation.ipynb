```python
### Change the datatypes

df['Value_new']=df['Value'].astype(float)
df.head()
```

```python
df['New_value'] = df['Value'].apply(lambda x:x*2)
df.head()
```

```python
###Data Aggregrating and Grouping 

df.head()
```

```python
grouped_mean = df.groupby('Product')['Value'].mean()
print(grouped_mean)
```

```python
grouped_sum=df.groupby(['Product','Region']).sum()
print(grouped_sum)
```

```python
grouped_m=df.groupby(['Product','Region'])['Value'].mean()
print(grouped_m)
```

```python
### Aggregrate multiple functions 

grouped_agg = df.groupby('Region')['Value'].aggregate(['mean','sum','count'])
print(grouped_agg)
```

```python
### Merging and joining Dataframes

df1 = pd.DataFrame({'Key': ['A','B','C'], 'Value': [1,2,3]})

df2 = pd.DataFrame({'Key': ['A','B','D'], 'Value': [4,5,6]})

```

```python
df1
```

```python
df2
```

```python
### Merge dataframe on the 'Key Columns' 

pd.merge(df1,df2,on="Key",how="inner")
```

```python
pd.merge(df1,df2,on="Key",how="outer")
```

```python
pd.merge(df1,df2,on="Key",how="left")
```

```python
pd.merge(df1,df2,on="Key",how="right")
```
